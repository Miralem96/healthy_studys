Kort förklaring på Del 2 Fördjupning och Pipeline

Vad gör pipeline koden?

1. Förbehandling av data
   tar bort saknade värden,
   gör om kategoriska variabler till numeriska,
   detta gör datan  ren och användbar för analys och modeller
   och maskininlärningsmodeler och statistiska metoder kräver rena, numeriska värden för att fungera korrekt

2. En enkel linjär regression
   jag testar hur en variabel i taget som påverkar bloodtrycket
   t.ex systolic_bp - age, där modellen ritar en linje över daetapunkterna och beräknar koefficient,intercept och r2 värde
   detta visar t.ex äldre personer tenderar att ha högre blodtryck

3. Multipel regression
   Här gör jag en modell som använder flera variabler samtidigt
   t.ex systolic_bp - age + weight
   Detta ger en tydligare bild av hur variablerna tillsammans förklarar variationen i blodtryck.

4. Residualanalys
   Här ritass ut residualer som är fel mellan verkligt värde och modellens värde.
   residualer visar om modellen passar bra eller om det finns mönster som modellen missar

5. Korrelationsmaatris
   Ritar en värmekarta som visar hur strakt variablerna hänger ihop
   Detta hjälper med att se vilka variabler som är viktiga att sudera vidare
